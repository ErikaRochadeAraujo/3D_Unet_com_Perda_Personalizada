```python
from transformers import pipeline

# Carregar o pipeline de análise de sentimentos
analisar_sentimento = pipeline("sentiment-analysis", model="distilbert-base-uncased-finetuned-sst-2-english")

# Lista de textos para analisar
textos = [
    "I love programming in Python, it's amazing!",
    "I hate when my code doesn't work, it's so frustrating.",
    "The weather today is neither good nor bad."
]

# Analisar o sentimento de cada texto
for texto in textos:
    resultado = analisar_sentimento(texto)[0]
    print(f"Texto: {texto}")
    print(f"Sentimento: {resultado['label']} (Confiança: {resultado['score']:.2f})")
    print("-" * 50)
```

```python
import requests
from bs4 import BeautifulSoup
from transformers import pipeline

# Função para extrair o texto de um site
def extrair_texto_do_site(url):
    # Fazendo a requisição HTTP
    headers = {
        "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36"
    }
    response = requests.get(url, headers=headers)

    # Verificando se a requisição foi bem-sucedida
    if response.status_code != 200:
        raise Exception(f"Erro ao acessar o site. Código de status: {response.status_code}")

    # Parseando o conteúdo HTML
    soup = BeautifulSoup(response.text, 'html.parser')

    # Extraindo o texto de todas as tags <p> (parágrafos)
    textos = [p.get_text() for p in soup.find_all('p')]

    # Juntando todos os textos em uma única string
    texto_completo = ' '.join(textos)

    return texto_completo

# Função para gerar um resumo do texto
def gerar_resumo(texto):
    # Carregar o pipeline de sumarização
    sumarizar = pipeline("summarization", model="facebook/bart-large-cnn")

    # Gerar o resumo (limitar o texto para evitar erros de tamanho)
    texto = texto[:1024]  # Limita o texto a 1024 caracteres
    resumo = sumarizar(texto, max_length=130, min_length=30, do_sample=False)

    return resumo[0]['summary_text']

# URL do site para extrair o texto
url = "https://www.bbc.com/news"  # Substitua pelo site que deseja analisar

# Extrair o texto do site
try:
    print("Extraindo texto do site...")
    texto = extrair_texto_do_site(url)
    print(f"\nTexto extraído do site:\n{texto[:500]}...\n")  # Exibe os primeiros 500 caracteres

    # Gerar o resumo do texto
    print("Gerando resumo...")
    resumo = gerar_resumo(texto)
    print(f"\nResumo do conteúdo:\n{resumo}")

except Exception as e:
    print(f"Erro: {e}")
```

```python
import streamlit as st
import requests
from bs4 import BeautifulSoup
from transformers import pipeline

# Função para extrair o texto de um site
def extrair_texto_do_site(url):
    headers = {
        "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36"
    }
    response = requests.get(url, headers=headers)
    if response.status_code != 200:
        raise Exception(f"Erro ao acessar o site. Código de status: {response.status_code}")
    soup = BeautifulSoup(response.text, 'html.parser')
    textos = [p.get_text() for p in soup.find_all('p')]
    texto_completo = ' '.join(textos)
    return texto_completo

# Função para gerar um resumo do texto
def gerar_resumo(texto):
    sumarizar = pipeline("summarization", model="facebook/bart-large-cnn")
    texto = texto[:1024]  # Limita o texto a 1024 caracteres
    resumo = sumarizar(texto, max_length=130, min_length=30, do_sample=False)
    return resumo[0]['summary_text']

# Interface do Streamlit
st.title("Resumidor de Conteúdo de Site")
url = st.text_input("Digite a URL do site:")

if url:
    try:
        st.write("Extraindo texto do site...")
        texto = extrair_texto_do_site(url)
        st.write("Texto extraído do site:")
        st.write(texto[:500] + "...")  # Exibe os primeiros 500 caracteres

        st.write("Gerando resumo...")
        resumo = gerar_resumo(texto)
        st.write("Resumo do conteúdo:")
        st.write(resumo)

    except Exception as e:
        st.error(f"Erro: {e}")
```
